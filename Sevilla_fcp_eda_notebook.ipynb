## ABC Budget Distribution

```python
plt.figure()
df['ABC'].plot(kind='hist')
plt.xlabel('Approved Budget for the Contract (ABC)')
plt.ylabel('Frequency')
plt.title('Distribution of ABC')
plt.show()
```

## Most projects have low to moderate ABC values, while a smaller number of projects have very high ABC values.

**This pattern is typical for infrastructure datasets because:**

    • Many projects are small canals, slope protections, desilting, drainage improvements
    • Only a few are major floodwalls, dikes, river improvements, or large structural works
## Most ABC values cluster at smaller amounts

**You will notice a large concentration in the lower bin(s):**

    •Most projects fall under roughly ₱5M – ₱50M

    •These are standard local flood mitigation projects

This cluster indicates that small-scale projects make up the majority of DPWH’s flood control spending.
## There are outliers, very large projects

**The long tail on the right side suggests:**

    •Some projects have ABCs in the hundreds of millions
    
    •These are large, capital-intensive flood control structures:
    
        •Major river embankments
        
        •Large dike systems
        
        •Flood gates and river improvements
        
        •Multi-component flood mitigation packages

These high-value projects are few but significantly drive total budget spending.

## Projects per InfraYear

```python
plt.figure()
df['InfraYear'].value_counts().sort_index().plot(kind='bar')
plt.xlabel('InfraYear')
plt.ylabel('Count')
plt.title('Projects by InfraYear')
plt.show()
```

## InfraYear corresponds closely to FundingYear

**Since InfraYear ≈ budget year, this tells us:**

    Budget allocations for flood control were higher in recent fiscal years

## Projects by Region

```python
plt.figure()
df['Region'].value_counts().plot(kind='bar')
plt.xlabel('Region')
plt.ylabel('Count')
plt.title('Projects by Region')
plt.show()
```

## Central Luzon is one of the most flood-prone regions in the Philippines

It has large river basins (Pampanga, Agno, Tarlac) that overflow frequently

Many low-lying plains make flooding recurrent

It contains highly populated cities like Angeles and San Fernando

**So, Region III naturally receives a big share of DPWH flood mitigation projects.**

## Region I, II, and V also show high project counts

These regions contain:

    Coastal areas affected by typhoons
    
    River deltas prone to overflow
    
    Mountain-to-lowland runoff that causes flooding

**They often rank high in project counts because flood control is a long-term, ongoing need.**

## Lower counts in Regions IX, XII, and XIII

These southern regions tend to have fewer projects in your dataset because:

    They have lower population density
    
    Less urbanized flood infrastructure
    
    Some areas rely more on natural river pathways than engineered structures
    
    DPWH reporting varies by region

**This doesn’t necessarily mean they have no flooding, only that the number of funded/recorded projects is fewer in this dataset.**

## ABC Distribution by Region (Boxplot)

```python
plt.figure(figsize=(10,6))
df.boxplot(column='ABC', by='Region', rot=90)
plt.title('ABC by Region')
plt.suptitle('')
plt.ylabel('Approved Budget for the Contract (ABC)')
plt.show()
```

**Central Tendency and Spread:** You can quickly compare the median ABC (the line in the box) and the overall spread (the length of the whiskers and the presence of outliers) for each region.

**Regional Differences:** Region III, Region IV-A, and Region V appear to have higher median ABCs and also show a substantial number of very high outliers (some approaching or exceeding $3 \times 10^8$).

**Outliers:** The plot clearly highlights that many regions have contracts with Approved Budgets that are significantly larger than the typical contracts in that region.

## Approved Budget for the Contract vs Contract Cost

```python
plt.figure()
plt.scatter(df['ABC'], df['ContractCost'], s=5)
plt.xlabel('Approved Budget for the Contract')
plt.ylabel('ContractCost')
plt.title('ABC vs Contract Cost')
plt.show()
```

**The Strong Diagonal Line:** Most of the dots cluster tightly along a diagonal line. This means that for nearly all contracts, the final cost is almost exactly the same as the approved budget.

**Contracts Below Budget:** You can see many dots falling below the main diagonal line. These dots represent contracts where the final cost was lower than the approved budget (ABC), suggesting money was saved during the bidding or negotiation process.

## Project Duration (Start to Actual Completion)

```python
df2 = df.copy()

df2['StartDate'] = pd.to_datetime(df2['StartDate'], errors='coerce')
df2['CompletionDateActual'] = pd.to_datetime(df2['CompletionDateActual'], errors='coerce')

df2['duration_days'] = (df2['CompletionDateActual'] - df2['StartDate']).dt.days

df2['duration_days'].plot(kind='hist', bins=50)
plt.xlabel('Duration (days)')
plt.title('Distribution of Project Durations')
plt.show()
```

**Skewed Distribution:** The most striking feature is that the distribution is highly right-skewed (or positively skewed). This means the majority of the data is concentrated on the left, but the "tail" stretches out to the right.

**Most Projects are Short:** The distribution peaks sharply around the 0 to 250 day mark. This tells us that the vast majority of projects in this dataset have a relatively short duration—less than a year (365 days).

**The Mode:** The tallest bar, or the mode, is located in the range of approximately 100 to 200 days. This is the most common project duration.

**Fewer Long Projects:** The frequency drops off very quickly.Projects lasting more than 500 days (around $1.4$ years) are much less common.Projects lasting 1,000 days (around $2.7$ years) or longer are rare, but they do exist, as shown by the very low bars extending out toward 2,000 days. These long-duration projects are often responsible for pulling the average duration higher, even though most projects are completed quickly.
